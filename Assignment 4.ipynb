```python
import math

class BankAccount:
    """Class representing a bank account with deposit, withdrawal, and string representation features."""
    def __init__(self, bankname, firstname, lastname, balance=0):
        self.bankname = bankname
        self.firstname = firstname
        self.lastname = lastname
        self.balance = balance

    def deposit(self, amount):
        """Allows the user to deposit money into their balance."""
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f} successfully.")
        else:
            print("Deposit amount must be greater than zero.")

    def withdrawal(self, amount):
        """Allows the user to withdraw money if sufficient balance is available."""
        if amount > self.balance:
            print(f"Withdrawal denied! Attempted to withdraw ${amount:.2f}, but available balance is ${self.balance:.2f}.")
        elif amount <= 0:
            print("Withdrawal amount must be greater than zero.")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f} successfully.")

    def __str__(self):
        """Returns string representation displaying bank name, owner name, and current balance."""
        owner_name = f"{self.firstname} {self.lastname}"
        return f"Bank Name: {self.bankname} | Owner: {owner_name} | Current Balance: ${self.balance:.2f}"


class Box:
    """Class representing a geometric box with various dimensions and calculation methods."""
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def render(self):
        """Prints out to the screen a box made with asterisks of length and width dimensions."""
        # Assuming length represents height (rows) and width represents columns (or vice versa)
        for _ in range(self.length):
            print("*" * self.width)

    def invert(self):
        """Switches length and width with each other."""
        self.length, self.width = self.width, self.length

    def get_area(self):
        """Returns the area of the box."""
        return self.length * self.width

    def get_perimeter(self):
        """Returns the perimeter of the box."""
        return 2 * (self.length + self.width)

    def double(self):
        """Doubles the size of the box (doubles both length and width)."""
        self.length *= 2
        self.width *= 2
        return self

    def __eq__(self, other):
        """Compares two boxes for equality based on identical length and width."""
        if isinstance(other, Box):
            return self.length == other.length and self.width == other.width
        return False

    def print_dim(self):
        """Prints to screen the length and width details of the box."""
        print(f"Box Dimensions -> Length: {self.length}, Width: {self.width}")

    def get_dim(self):
        """Returns a tuple containing the length and width of the box."""
        return (self.length, self.width)

    def combine(self, other):
        """Takes another box as an argument and increases length and width by the dimensions passed in."""
        if isinstance(other, Box):
            self.length += other.length
            self.width += other.width

    def get_hypot(self):
        """Finds the length of the diagonal that cuts through the middle."""
        return math.sqrt(self.length**2 + self.width**2)


# ==========================================
# TESTING AND EXECUTION ACCORDING TO RUBRIC
# ==========================================

if __name__ == "__main__":
    print("--- Q1: BankAccount Testing ---")
    my_account = BankAccount("Chase", "Jane", "Doe", 100)
    print(my_account)
    my_account.deposit(250)
    my_account.withdrawal(50)
    my_account.withdrawal(500)  # Should fail due to insufficient balance
    print(my_account)
    print("\n" + "="*40 + "\n")

    print("--- Q2: Box Class Testing and Modifications ---")
    # Instantiate 3 boxes of dimensions (5,10), (3,4), and (5,10)
    box1 = Box(5, 10)
    box2 = Box(3, 4)
    box3 = Box(5, 10)

    # Print dimension info for each using print_dim()
    print("Box 1:")
    box1.print_dim()
    print("Box 2:")
    box2.print_dim()
    print("Box 3:")
    box3.print_dim()

    # Evaluate if box1 == box2, and evaluate if box1 == box3, print True or False
    print(f"\nIs box1 == box2? {box1 == box2}")
    print(f"Is box1 == box3? {box1 == box3}")

    # Combine box3 into box1 (i.e., box1.combine(box3))
    print("\nCombining box3 into box1...")
    box1.combine(box3)
    print("New dimensions of box1 after combining box3:")
    box1.print_dim()

    # Double the size of box2
    print("\nDoubling the size of box2...")
    box2.double()
    print("New dimensions of box2 after doubling:")
    box2.print_dim()

    # Combine box2 into box1
    print("\nCombining box2 into box1...")
    box1.combine(box2)
    print("Final dimensions of box1 after combining box2:")
    box1.print_dim()
```
